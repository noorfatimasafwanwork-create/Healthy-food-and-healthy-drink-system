```python
# ===============================
# Healthy Food & Drink Recommendation System
# Using Pandas (School Project) MADE BY AI - CHATGPT NOT BY ME
# ALLOWED AI BY MISS HIBA  - and, I did change the code a little bit.
# ===============================

# 1. Import Libraries
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import NearestNeighbors

# -------------------------------
# 2. Data Collection
# -------------------------------
data = {
    "item_name": [
        "Apple", "Orange Juice", "Grilled Chicken", "Salad",
        "Brown Rice", "Yogurt", "Green Tea", "Orange Juice"
    ],
    "type": [
        "food", "food", "food", "food",
        "food", "food", "drink", "drink"
    ],
    "calories": [52, 96, 165, 33, 123, 59, 2, 45],
    "protein": [0.3, 1.3, 31, 2, 2.7, 10, 0, 0.7],
    "sugar": [10, 12, 0, 2, 0.4, 4, 0, 8],
    "fat": [0.2, 0.3, 3.6, 0.4, 1, 0.4, 0, 0],
    "is_healthy": [1, 1, 1, 1, 1, 1, 1, 0]
}

df = pd.DataFrame(data)

# -------------------------------
# 3. User Registration & Preferences - Kind of edited
# -------------------------------
user_profile = {
    "age": 15,
    "height_cm": 155,
    "weight_kg": 55,
    "activity_level": "moderate",
    "goal": "weight_loss",
    "dietary_preference": "vegetarian",
    "allergies": []
}

# -------------------------------
# 4. Data Cleaning & Preprocessing
# -------------------------------
# Remove duplicate records
df = df.drop_duplicates()

# Handle missing values
df = df.fillna(0)

# Select nutritional features
features = ["calories", "protein", "sugar", "fat"]

# Scale numerical features
scaler = StandardScaler()
scaled_features = scaler.fit_transform(df[features])

# -------------------------------
# 5. Feature Selection
# -------------------------------
X = scaled_features
y = df["is_healthy"]

# -------------------------------
# 6. Model Selection & Training
# -------------------------------
model = NearestNeighbors(n_neighbors=3, metric="euclidean")
model.fit(X)

# -------------------------------
# 7. Model Evaluation
# -------------------------------
healthy_ratio = len(df[df["is_healthy"] == 1]) / len(df)
print(f"Model Accuracy (Healthy Item Ratio): {healthy_ratio:.2f}")

# -------------------------------
# 8. Recommendation System
# -------------------------------
def recommend_healthy_items(df, model, scaler):
    # Filter for healthy items to get their scaled features for querying
    healthy_items_for_query = df[df["is_healthy"] == 1]
    features = ["calories", "protein", "sugar", "fat"]

    # Scale the features of healthy items for querying
    scaled_data_for_query = scaler.transform(healthy_items_for_query[features])

    # Find k-nearest neighbors from the *full* dataset (model trained on X)
    # The `indices` returned are relative to the *full* original dataframe `df`
    distances, indices = model.kneighbors(scaled_data_for_query)

    # The original error occurred because `healthy_df.iloc[indices[0]]` used indices
    # valid for `df` but not necessarily for `healthy_df`.
    # We now retrieve from the *full* dataframe `df` using these indices.
    raw_recommendations = df.iloc[indices[0]]

    # Filter these raw recommendations to only include healthy items,
    # ensuring the function's name is accurately reflected.
    healthy_recommendations = raw_recommendations[raw_recommendations["is_healthy"] == 1]

    # Return the specified columns from the healthy recommendations
    return healthy_recommendations[["item_name", "type", "calories"]]

# -------------------------------
# 9. Output & Results
# -------------------------------
recommendations = recommend_healthy_items(df, model, scaler)

print("\nRecommended Healthy Foods & Drinks:")
for item in recommendations["item_name"]:
    print("-", item)
```
